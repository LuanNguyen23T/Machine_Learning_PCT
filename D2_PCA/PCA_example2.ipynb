### **Bước 1**:

```python
# Import Required Libraries
import numpy as np
from matplotlib import pyplot as plt
# Plot the points
Points = np.array([[6, 6], [4, 4], [4, 6], [6, 4], [7, 7], [3, 3], [3, 7], [7, 3]])
Points
```

```python
plt.plot(Points[:, 0], Points[:, 1], 'rd')
plt.title('Original Points')
plt.show()
```

### **Bước 2**:

```python
Mean = np.mean(Points, axis=0)   #tính mean
Points_Norm = Points - Mean         #x-mean
plt.plot(Points_Norm[:, 0], Points_Norm[:, 1], 'bx')
plt.title('Normalized Points')
plt.show()
```

```python
row, col = Points_Norm.shape
Sample_Cov_Matrix = np.matmul(Points_Norm.T, Points_Norm)/(row - 1) # ma trận hiệp phương sai
Eigen_Values, Eigen_Vectors = np.linalg.eig(Sample_Cov_Matrix)   # tinh giá trị riêng, vector riêng của ma trận hiệp phương sai
```

```python
Significance = [np.abs(i)/np.sum(Eigen_Values) for i in Eigen_Values]
K = len(Eigen_Values)
x = np.arange(1, K+1)
y = np.cumsum(Significance)
plt.plot(x, y)
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Variance (%)')
plt.show()
```

```python
# Scree Plot showing Variance of each Principal Component
plt.plot(x, Significance)
plt.xlabel('Number of Components')
plt.ylabel('Variance (%)') # for each component
plt.show()
```
