```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
```

```python
X = np.random.rand(10,10)  # 2D ndarray
X
```

```python
n_samples, n_features = X.shape
#n_components_ = n_components or n_features
```

```python
 #Chuẩn hóa các tính năng bằng cách loại bỏ giá trị trung bình và chia tỷ lệ thành phương sai đơn vị
 X_std = (X - np.mean(X, axis=0)) / np.std(X, axis=0)
 X_std
```

```python
cov_mat = np.corrcoef(X, rowvar=False)
print('Covariance matrix: \n%s' % cov_mat)
```

```python
eig_vals, eig_vecs = np.linalg.eig(cov_mat)
eig_vals
```

```python
eig_vecs
```

```python
idx = eig_vals.argsort()[::-1]
eig_vals, eig_vecs = eig_vals[idx], eig_vecs[:, idx]
```

```python
# sắp xếp các giá trị riêng theo thứ tự giảm dần
k=5
print("Eigenvalues:\n", eig_vals[:k])
        # a matrix of eigenvectors (each column is an eigenvector)
print("Eigenvectors:\n", eig_vecs[:k])
```

```python
explained_variance = (eig_vals / np.sum(eig_vals))[:k]
explained_variance_cumsum = np.cumsum(explained_variance)
total_var = np.sum(explained_variance)
```

```python
print('Total variance explained:\n', total_var)
print('Explained variance:\n', explained_variance)
print('Explained variance cumsum:\n', explained_variance_cumsum)
```

```python
# Projection matrix and dimensionality reduction
W = eig_vecs[:, 0:k]
#assert W.shape == (n_features, k)
```

```python
PC_k = X.dot(W)
#assert PC_k.shape == (n_samples, k)
```
