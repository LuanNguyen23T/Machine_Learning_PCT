## Cấu trúc fuzzy c-means

Tìm minimum và maximum

```python
def find_max(dataset):
  cols = dataset.shape[1]
  max_in_dataset = -sys.maxsize
  for col in range(cols):
    col_max = np.max(dataset[:, col])
    if col_max > max_in_dataset:
      max_in_dataset = col_max
  return max_in_dataset
```

```python
def find_min(dataset):
  cols = dataset.shape[1]
  min_in_dataset = sys.maxsize
  for col in range(cols):
    col_min = np.min(dataset[:, col])
    if col_min < min_in_dataset:
      min_in_dataset = col_min
  return min_in_dataset
```

Chọn các giá trị random

```python
def initialize_centroids(dataset, k, minimum=None, maximum=None):
  cols = dataset.shape[1]
  centroids = np.zeros((k, cols))
  for col in range(cols):
    col_min = np.min(dataset[:, col])
    col_max = np.max(dataset[:, col])
    centroids[:, col] = np.random.uniform(col_min, col_max, k)
  return centroids
```

Tính độ mờ cho mỗi điểm
<br>
u.png

```python
def update_data_fuzziness(dataset, centroids, m):
  eps = 1e-9
  u = []
  for data in dataset:
    dists = np.array([np.linalg.norm(data - vi) for vi in centroids])
    if np.any(dists < eps):
      zero_idx = np.argmin(dists)
      row = np.zeros(len(centroids))
      row[zero_idx] = 1.0
      u.append(row)
    else:
      row = [1.0 / np.sum([(dists[i] / (dists[j] + eps)) ** (2 / (m - 1)) for j in range(len(centroids))]) for i in range(len(centroids))]
      u.append(row)
  return np.array(u, dtype='float64')
```

Tìm giá trị trung bình của các điểm
<br>

image.png

```python
def modify_centroids(num_centroids, u, dataset, m):
  N = len(dataset)
  eps = 1e-9
  centroids = np.array([
    np.add.reduce([(u[k, i] ** m) * dataset[k] for k in range(N)]) / (np.sum([u_[i] ** m for u_ in u]) + eps) 
    for i in range(num_centroids)
  ])
  return centroids
```

Tính độ sai số

```python
def error_calculator(u, m, dataset, centroids):
  error = np.sum([np.sum([(u[j, i] ** m) * (np.linalg.norm(dataset[j] - centroids[i]) ** 2) 
            for i in range(len(centroids))]) 
      for j in range(len(dataset))])
  return error
```

Tạo model FCM

```python
def fuzzy_cmeans_model(dataset, k, num_iterations, m):
  minimum = find_min(dataset)
  maximum = find_max(dataset)

  centroids = initialize_centroids(dataset, k, minimum, maximum)

  for i in range(num_iterations):
    u = update_data_fuzziness(dataset, centroids, m)
    centroids = modify_centroids(len(centroids), u, dataset, m)
    error = error_calculator(u, m, dataset, centroids)

  return centroids, u, error
```

Plot kết quả

```python
def plot_results(dataset, u, centroids, k):
  plt.figure(figsize=(8,6))
  colors = plt.cm.tab10(np.linspace(0, 1, k))
  for cluster in range(k):
    clusteredData = np.array([dataset[i] for i in range(len(dataset)) 
        if cluster == np.argmax(u[i])])
    if len(clusteredData) > 0:
      plt.scatter(clusteredData[:, 0], clusteredData[:, 1], color=colors[cluster], label=f'Cluster {cluster + 1}')
  plt.scatter(centroids[:, 0], centroids[:, 1], color='black', s=120, marker='X', label='Centroids')
  plt.title('Fuzzy C-Means Clustering Results')
  plt.xlabel('Petal Length (cm)')
  plt.ylabel('Petal Width (cm)')
  plt.legend()
  plt.show()
```

```python
def fuzzy_plot(dataset, u, centroids, k):
  plt.figure(figsize=(8,6))
  if k == 3:
    # Với k=3, độ thuộc (u1, u2, u3) được ánh xạ trực tiếp thành màu RGB
    plt.scatter(dataset[:, 0], dataset[:, 1], c=u, s=60, edgecolors='k', linewidth=0.5)
  else:
    labels = np.argmax(u, axis=1)
    plt.scatter(dataset[:, 0], dataset[:, 1], c=labels, cmap='viridis', s=60, edgecolors='k', linewidth=0.5)
  plt.scatter(centroids[:, 0], centroids[:, 1], color='red', s=150, marker='X', label='Centroids')
  plt.title(f'Fuzzy Membership Plot (k={k})')
  plt.xlabel('Petal Length (cm)')
  plt.ylabel('Petal Width (cm)')
  plt.legend()
  plt.show()
```

## Áp dụng vào dataset

```python
errors = []

for i in range(1,7):
  centroids, u, error = fuzzy_cmeans_model(dataset=data1, k=i, num_iterations=100, m=2)
  errors.append(error)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(np.arange(1, 7), errors, marker='o', color='b', linestyle='-')
plt.title('Errors with different number of clusters for Iris Dataset')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Objective Function Error')
plt.grid(True)
plt.show()
```
